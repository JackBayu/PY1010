## 2.0 Oppgåve 2) - Priskalkulerering for elevar

```python
import math

mengde_elevar = int(input('Skriv mengde elevar:'))
print()

brøkdel = 1/4
innkjøp_pizza = math.ceil(mengde_elevar*brøkdel)

print(f"For {mengde_elevar} elevar, skal det kjøpast inn {innkjøp_pizza} heile pizza.")
```

### 2.0.1 Forklaring av løysingsforslag oppgåve 2)

Programmet oppfyller krava i oppgåva ved at brukaren skriv inn talet på elevar med funksjonen input(), og verdien blir konvertert til heiltal med int(). Deretter blir mengda pizza rekna ut ved å multiplisere talet på elevar med 1/4.

Eg vurderte først å bruke round() for å runde resultatet til næraste heiltal. Etter å ha lese oppgåveteksten og vurdert hintet om 21 elevar, såg eg at dette ikkje alltid ville gi nok pizzaer. Sidan ein må kjøpe heile pizzaer, valde eg å bruke ceil() frå math-biblioteket. Denne funksjonen rundar alltid opp til næraste heiltal, slik at det blir kjøpt inn nok pizzaer til alle elevane.

# 3.0 Oppgåve 3) - Rekna om frå grader til radianar

## 3.1 Alternativ 1 i samsvar med oppgåva

```python
import numpy as np
import sympy as sp
from IPython.display import display, Math

def v_rad(grader):
    v_rad = grader * np.pi / 180
    return v_rad

v_grad = float(input("Skriv inn gradtalet: "))

print(f"{v_grad}° = {round(v_rad(v_grad), 2)} rad")
```

## 3.2 Alternativ 2 med passande tekst og verdi 

<img src= "einingssirkel_visuelt.jpg" width="400">

```python
v_rad_sympy = sp.Rational(v_grad) * sp.pi / 180

display(Math(f"{v_grad}^\\circ = {sp.latex(v_rad_sympy)}"))
```

### 3.0.1 Forklaring av løysingsforslag oppgåve 3

Brukaren skriv inn gradtalet med `input()`, og verdien blir konvertert til `float`. I alternativ 1 brukar eg ein funksjon som reknar om grader til radianar ved hjelp av `numpy` og formelen:

$$v_{rad}=v_{grad}\cdot\frac{\pi}{180}$$

I alternativ 2 brukar eg sympy for å vise radianverdien i eksakt matematisk form. `Rational()` gjer det mogleg å uttrykkje resultatet som ein brøk med $\pi$. `display()` og `Math()` viser uttrykket matematisk i Jupyter Notebook, medan `sp.latex()` konverterer uttrykket til LaTeX-format.

## 5.0 Oppgåve 5 – Utrekning av areal og ytre omkrins

<img src= "figurtiloppgaave.jpg" width="200">

#### $a$ = breidde / hostliggjande katet 
#### $b$ = høgde / motståande katet
#### $r$ = radius / 2 $\div$ diameter($d$)
#### $ \angle \alpha$ = vinkel alfa
####
#### Areal til ein halvsirkel:
$A = \frac{\pi r^2}{2}$

#### Omkrins til ein halvsirkel:
$O = \pi r + 2r$

#### Areal til ein rettvinkla trekant:
$\triangle A = \frac{a \cdot b}{2}$

#### Omkrins til ein rettvinkla trekant:
$\triangle O = a + b + \sqrt{a^2 + b^2}$

#### Areal til heile figuren:
$$A = \frac{\pi r^2 + a \cdot b}{2}$$

#### Omkrins til heile figuren:
$$O = \pi r + b + \sqrt{a^2 + b^2}$$


```python
import math

def figur(a, b):     # Funksjon som reknar ut areal og ytre omkrins til figuren
    r = a / 2        # Radiusen er halvparten av diameteren a

    areal_halvsirkel = math.pi * r**2 / 2    # Reknar ut arealet til halvsirkelen
    areal_trekant = a * b / 2                # Reknar ut arealet til den rettvinkla trekanten
    areal = areal_halvsirkel + areal_trekant # Legg saman areala

    hypotenus = math.sqrt(a**2 + b**2)       # Reknar ut hypotenusen med Pytagoras' setning
    omkrins = math.pi * r + b + hypotenus    # Reknar ut den ytre omkrinsen til figuren

    return areal, omkrins                    # Returnerer areal og omkrins
    
areal, omkrins = figur(5, 5)                 # Her blir verdiane 5 og 5 sendte inn som argument til funksjonen figur()

print(f"Areal: {areal:.2f}")
print(f"Ytre omkrins: {omkrins:.2f}")
```

## 6.0 Plotta funksjon på eit interval

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10, 10, 200) # Lager 200 punkt frå -10 til 10
f = -x**2 - 5                 # Definerer funksjonen

plt.plot(x, f)                      # Teiknar grafen

plt.title("Graf av f(x) = -x² - 5") # Legg til tittel og aksenamn
plt.xlabel("x")
plt.ylabel("f(x)")

plt.grid()                          # Viser rutenett
plt.show()                          # Viser plottet
```
