## 1.0 Oppgåve 1) - Laga eit program for inntasting av alder

```python
år = 2026
fødselsår = int(input('Kva for år blei du født?'))
print()

print(f"Vi er no i {år}, og eg er {år - fødselsår} år gammal.")

dagens_alder = år - fødselsår

print(f"Neste år i {år+1}, blir eg {dagens_alder + 1} år gammal.")
```

### 1.0.1 Forklaring av løysingsforslag oppgåve 1)

Programmet oppfyller minimumskrava i oppgåva ved at brukaren skriv inn fødselsåret sitt med funksjonen `input()`, og verdien blir konvertert til heiltal med `int()`. Deretter blir alderen rekna ut ved å trekke fødselsåret frå det valde året (2026 i denne løysinga). Resultatet blir så skrive ut på skjermen med forklarande tekst.

Eg har valt å bruke strengar (strings) og f-strengar (`f-strings`) for å presentere resultatet på ein lesbar måte. F-strengar gjer det mogleg å setje inn verdiar frå variablar direkte i teksten utan ekstra samansetjing av strengar. Dette gir ein enklare og meir oversiktleg kode for tekstbehandling og utskrift.

Valet av f-strengar er basert på informasjon frå KI og litteraturen til Gaddis (2019), der f-strengar blir presenterte som ein effektiv metode for formatering av tekst og utskrift i Python.

## 2.0 Oppgåve 2) - Priskalkulerering for elevar

```python
import math

mengde_elevar = int(input('Skriv mengde elevar:'))
print()

brøkdel = 1/4
innkjøp_pizza = math.ceil(mengde_elevar*brøkdel)

print(f"For {mengde_elevar} elevar, skal det kjøpast inn {innkjøp_pizza} heile pizza.")
```

### 2.0.1 Forklaring av løysingsforslag oppgåve 2)

Programmet oppfyller krava i oppgåva ved at brukaren skriv inn talet på elevar med funksjonen input(), og verdien blir konvertert til heiltal med int(). Deretter blir mengda pizza rekna ut ved å multiplisere talet på elevar med 1/4.

Eg vurderte først å bruke round() for å runde resultatet til nærmaste heiltal. Etter å ha vurdert oppgåveteksten, valde eg i staden å bruke ceil() frå math-biblioteket. Dette rundar alltid resultatet opp til næraste heiltal, slik at det blir kjøpt inn nok heile pizzaer.

# 3.0 Oppgåve 3) - Rekna om frå grader til radianar

<img src= "einingssirkel_formel.jpg" width="400">

## 3.1 Alternativ 1 i samsvar med oppgåva

```python
import numpy as np
import sympy as sp
from IPython.display import display, Math

v_grad = float(input("Skriv inn gradtalet: "))

v_rad = v_grad * np.pi / 180

print(f"{v_grad}° = {round(v_rad, 2)} rad")
```

## 3.2 Alternativ 2 med passande tekst og verdi 

<img src= "einingssirkel_visuelt.jpg" width="400">

```python
v_grad = float(input("Skriv inn gradtalet: "))

v_rad = sp.Rational(v_grad) * sp.pi / 180

display(Math(f"{v_grad}^\\circ = {sp.latex(v_rad)}"))
```

### 3.0.1 Forklaring av løysingsforslag oppgåve 3

Programmet oppfyller krava i oppgåva ved at brukaren skriv inn gradtalet med funksjonen input(), og verdien blir konvertert til float. Gradane blir deretter rekna om til radianar ved hjelp av formelen:

$$ v_{rad}=v_{grad}\cdot\frac{\pi}{180} $$

Eg har valt å bruke numpy for pi, sidan dette er biblioteket som er lagt opp til i oppgåva. I tillegg har eg brukt sympy for å kunne behalde den eksakte matematiske forma på resultatet. Ved å bruke Rational() kan resultatet uttrykkjast som ein brøk med \(\pi\), i staden for eit avrunda desimaltal. Til dømes blir 30 grader uttrykt som \(\frac{\pi}{6}\).

display() og Math() frå IPython.display blir brukt for å vise resultatet som matematisk notasjon i Jupyter Notebook. sp.latex() gjer om resultatet frå sympy til LaTeX-format.

I utskrifta blir ^\\circ brukt for å vise gradteiknet som ein del av LaTeX-notasjonen. Dette gjer at resultatet blir presentert matematisk som:

$$ 30^\circ=\frac{\pi}{6} $$

På denne måten utfører programmet både sjølve omrekninga og viser resultatet i eksakt matematisk form.
